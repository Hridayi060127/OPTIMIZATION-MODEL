# Optimization Using Linear Programming (PuLP)

## Product Mix Business Problem
A company produces Product A and Product B.  
The goal is to maximize profit under limited resources.

```python
!pip install pulp
```

```python
from pulp import *

# Create model
model = LpProblem("Maximize_Profit", LpMaximize)

# Variables
x = LpVariable("A", lowBound=0)
y = LpVariable("B", lowBound=0)

# Objective
model += 40*x + 30*y

# Constraints
model += 2*x + y <= 100
model += x + y <= 80

# Solve
model.solve()

# Output
print("Product A:", x.value())
print("Product B:", y.value())
print("Max Profit:", value(model.objective))
```

Result:
Product A = 20 units
Product B = 60 units
Maximum Profit = ₹2600
Explanation:
Product B is produced more because it uses fewer machine hours
Labor is fully used, so it is the limiting factor
The company should focus more on Product B to maximize profit.

```python
# Increase labor capacity
model = LpProblem("Maximize_Profit_New", LpMaximize)

x = LpVariable("A", lowBound=0)
y = LpVariable("B", lowBound=0)

model += 40*x + 30*y
model += 2*x + y <= 100
model += x + y <= 100   # increased labor

model.solve()

print("New Product A:", x.value())
print("New Product B:", y.value())
print("New Profit:", value(model.objective))
```

What-if Analysis Result:
Product A = 0 units
Product B = 100 units
Maximum Profit = ₹3000
Explanation:
After increasing labor, the company is no longer restricted by labor
Product B becomes the best option because:
It uses fewer machine hours
It gives good profit with less resource usage
The model chooses to produce only Product B to maximize profit

##Conclusion

By using limited resources efficiently, the company can maximize profit.
When constraints change (like increasing labor), the optimal strategy also changes.
This demonstrates the importance of resource planning in real-world business problems.
